**Research first; report writing later.** This notebook analyzes the official data and checks the two research questions. No observations are generated.

## Step 1 — Load packages and configure official dataset
Place the official `waterquality.csv` in this folder. This notebook depends on the fully documented functions in `analyze_waterquality.py`.

```python
from pathlib import Path
import pandas as pd
from IPython.display import display, Markdown, Image
import analyze_waterquality as aw
DATA_FILE = Path("waterquality.csv")
RESULTS_DIR = Path("results")
print("Dataset exists:", DATA_FILE.exists())
if not DATA_FILE.exists():
    print("Download official CSV from:", aw.DATA_URL)
```

## Step 2 — Validate sample sizes and data quality
Verify 900 rows, five river sections, four seasons, and 45 measurements in every section–season cell. Preserve repeated recorded values unless there is evidence of an entry error.

```python
data, audit = aw.load_data(DATA_FILE)
display(data.head(10))
print("Original rows:", audit["source_rows"], "complete rows:", audit["complete_rows"])
print("Missing values:", audit["missing_values_after_conversion"])
print("Warnings:", audit["warnings"])
display(audit["cell_counts"].pivot(index="section", columns="season", values="n"))
```

## Step 3 — Generate all descriptive, inferential and diagnostic results
Tests: Welch spring–autumn contrasts (Holm correction for two outcomes); Welch one-way ANOVA and Games–Howell section post-hoc testing; two-factor season × section ANOVA; assumption diagnostics.

```python
analysis = aw.run_analysis(DATA_FILE, RESULTS_DIR)
```

## Step 4 — Inspect tables directly

```python
for title, name in [("Spring versus autumn", "spring_vs_autumn"),
                    ("Section differences", "section_omnibus"),
                    ("Pairwise sections", "section_posthoc"),
                    ("Factorial model", "factorial_anova"),
                    ("Assumption diagnostics", "diagnostics")]:
    display(Markdown("### " + title))
    display(analysis[name])
```

## Step 5 — Review report-ready figures

```python
for outcome in aw.OUTCOMES:
    for figure in ["season_boxplot", "section_boxplot", "interaction", "residuals_qq"]:
        display(Markdown(f"### {outcome.title()}: {figure.replace('_', ' ')}"))
        display(Image(filename=str(RESULTS_DIR / f"figure_{outcome}_{figure}.png"), width=650))
```
